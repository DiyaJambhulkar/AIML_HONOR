```python
from numba import njit,prange
```

```python
@njit(parallel=True,fastmath=True)
def monte_carlo_risk(n_simulations:int)->float:
    acc=0
    for i in prange(n_simulations):
        x=np.random.random()
        y=np.random.random()
        if(x **2 + y**2) <=1.0:
            acc+=1
            return 4.0 * acc/n_simulations
        
        
```

```python
monte_carlo_risk(30000000)
```
