```python
import numpy as np
import time
from numba import njit, prange

# Step 1: Profiling decorator
def profiling(func):
    def wrapper(n):
        start = time.time()
        result = func(n)
        end = time.time()

        print("Time taken:", end - start, "seconds")
        return result

    return wrapper


# Step 2: Normal Python function
@profiling
def normal_function(n_simulations):
    acc = 0

    for i in range(n_simulations):
        x = np.random.random()
        y = np.random.random()

        if x**2 + y**2 <= 1.0:
            acc += 1

    return 4.0 * acc / n_simulations


# Step 3: Numba function
@njit(parallel=True, fastmath=True)
def monte_carlo_risk(n_simulations):
    acc = 0

    for i in prange(n_simulations):
        x = np.random.random()
        y = np.random.random()

        if x**2 + y**2 <= 1.0:
            acc += 1

    return 4.0 * acc / n_simulations


@profiling
def numba_function(n_simulations):
    return monte_carlo_risk(n_simulations)


# Step 4: Compare execution time
n = 1000000

# Compile Numba function before measuring its execution time
monte_carlo_risk(10)

print("Normal Python Function:")
print("Pi Value:", normal_function(n))

print("\nNumba Function:")
print("Pi Value:", numba_function(n))
```
